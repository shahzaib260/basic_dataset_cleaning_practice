```python
import numpy as np
import pandas as pd
```

```python
data = {
    'EmployeeID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack'],
    'Age': [25, 32, np.nan, 28, 45, 38, np.nan, 22, 41, 36],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR', 'Finance', 'IT'],
    'Salary': [50000, 60000, 55000, 58000, 72000, 68000, 59000, 47000, 71000, 65000],
    'Bonus%': [5, 10, np.nan, 0, 15, 8, 12, np.nan, 20, 0],
    'JoiningMonth': ['January', 'March', 'February', 'January', 'April', 'March', 'January', 'May', 'June', 'January']
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
```

```python
df
```

```python
df['Bonus%'].replace(np.nan, df['Bonus%'].mean(), inplace=True)
```

```python
df
```

```python
df.loc[(df['Bonus%'] == 0), 'GetBonus'] = 'No'
df.loc[(df['Bonus%'] > 0), 'GetBonus'] = 'Yes'
```

```python
df.loc[(df['Age'] <= 25), 'AgeGroup'] = 'Young'
df.loc[(df['Age'] > 25) & (df['Age'] < 40), 'AgeGroup'] = 'Mid Age'
df.loc[(df['Age'] >= 40), 'AgeGroup'] = 'Senior'
```

```python
df
```

```python
def slice_month(value):
    return value[:3]
```

```python
df['JoiningMonth'] = df['JoiningMonth'].map(slice_month)
```

```python
df['Age'] = df['Age'].astype(int)
```

```python
df
```

```python
gp1  = df.groupby(['Department', 'Name']).agg({'Salary': 'sum'})
```
